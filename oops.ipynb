```python
class Human:
    aadhar_no = "" # Class property 
    name= "" # Class property
    age = 0 # Class property

    def __init__(self, aadhar, name, age, city):
        self.ins_name = "Ducat India" # Instance property
        self.setData(aadhar, name, age, city)
        
    def getData(self):
        print("Aadhar no. =",self.aadhar_no)
        print("Name =",self.name)
        print("Age =",self.age)
        print("Institute name = ",self.ins_name)
        print("City = ",self.city)
        

    def setData(self, aadhar, name, age, city):
        self.aadhar_no = aadhar
        self.name = name
        self.age = age
        self.city = city
        self.getData()
```
