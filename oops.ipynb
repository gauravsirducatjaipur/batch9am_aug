```python
class Human:
    aadhar_no = ""
    name= ""
    age = 0

    def __init__(self):
        # print("constructor called successfully")
        self.setData()

    def getData(self):
        print("aadhar no. =",self.aadhar_no)
        print("name =",self.name)
        print("age =",self.age)
        

    def setData(self):
        self.aadhar_no = input("Enter your aadhar number ")
        self.name = input("Enter your name ")
        self.age = input("Enter your age ")
        self.getData()
```
