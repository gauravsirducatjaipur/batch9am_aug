```python
x = 10
try:
  print(x)
except:
  print("Something went wrong")
else:
  print("Nothing went wrong")
finally:
  print("It will execute always either exception or not")
```

```python
try:
  f = open("demofile.txt")
  try:
    f.write("Hello user")
  except:
    print("Something went wrong when writing to the file")
  finally:
    f.close()
except:
  print("Something went wrong when opening the file")
```

```python
try:
    num = int(input("Enter a number: "))

    if num > 0:
        print("Positive number")
    elif num < 0:
        print("Negative number")
    else:
        print("Zero")

except:
    print("It will handle the exception")
else:
    print("It will execute if there is no exception")
finally:
    print("It will execute always")
```
